QUESTION 1: BFS + Shortest Path

Topics: BFS + Shortest Path

Task:
Write a function that uses BFS to find the shortest path length from a given source node to all other nodes in an unweighted directed graph.

Input:

graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D', 'E'],
    'D': ['F'],
    'E': [],
    'F': []
}
start = 'A'


Output:

A : 0
B : 1
C : 1
D : 2
E : 2
F : 3

```python
from collections import deque

def BFS(graph,start):
    distance={}
    queue=deque([start])
    distance[start]=0
    while queue:
        node=queue.popleft()
        for neighbor in graph[node]:
            if neighbor not in distance:
                distance[neighbor]=distance[node]+1
                queue.append(neighbor)
    return distance           

graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D', 'E'],
    'D': ['F'],
    'E': [],
    'F': []
}
start = 'A'
result=BFS(graph,start)
for node in result:
    print(node,":",result[node])
```

QUESTION 2: DFS + Connected Graph

Topics: DFS + Connected Graph

Task:
Write a program that checks whether a given graph is connected using DFS.

Input:

graph = {
    0: [1, 2],
    1: [0, 2],
    2: [0, 1, 3],
    3: [2]
}


Output:

Graph is Connected

```python
def DFS(graph,node,visited):
    visited.add(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            DFS(graph,neighbor,visited)
def is_connected(graph):
    visited=set()
    start_node=next(iter(graph))
    DFS(graph,start_node,visited)
    if len(visited)==len(graph):
        print("Graph is connected")
    else:
        print("Graph is not connected")
graph = {
    0: [1, 2],
    1: [0, 2],
    2: [0, 1, 3],
    3: [2]
}
is_connected(graph)                        
```

QUESTION 3: BFS + Bipartite Graph

Topics: BFS + Bipartite Graph

Task:
Check whether the given undirected graph is bipartite using BFS coloring.

Input:

graph = {
    0: [1, 3],
    1: [0, 2],
    2: [1, 3],
    3: [0, 2]
}


Output:

Graph is Bipartite

```python
from collections import deque

def is_bipartite(graph):
    color={}
    for start in graph:
        if start not in color:
            queue=deque([start])
            color[start]=0
            while queue:
                node=queue.popleft()
                for neighbor in graph[node]:
                    if neighbor not in color:
                        color[neighbor]=1-color[node]
                        queue.append(neighbor)
                        if color[neighbor]==color[node]:
                            print("Graph is not Bipartite")
                            return
    print("Graph is Bipartite")
graph = {
    0: [1, 3],
    1: [0, 2],
    2: [1, 3],
    3: [0, 2]
}
is_bipartite(graph)
```

QUESTION 4: DFS + Longest Path (DAG only)

Topics: DFS + Longest Path

Task:
Given a Directed Acyclic Graph (DAG), find the longest path length from a given source node.

Input:

graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D'],
    'D': []
}
start = 'A'


Output:

Longest Path Length: 2

```python
def longest_path_dag(graph,node,memo): #memo=memoization
    if node in memo:
        return memo[node]
    if not graph[node]:
        memo[node]=0
        return 0
    max_length=0
    for neighbor in graph[node]:
        length=1+longest_path_dag(graph,neighbor,memo)
        max_length=max(max_length,length)
    memo[node]=max_length
    return max_length
graph={
    "A":["B","C"],
    "B":["D"],
    "C":["D"],
    "D":[]
    }
start="A"
memo={}
result=longest_path_dag(graph,start,memo)
print("Longest Path Length: ",result)
```

QUESTION 5: BST Creation + Inorder Traversal

Topics: Binary Search Tree + Inorder Sequence

Task:
Create a Binary Search Tree and print its Inorder traversal.

Input:

Enter elements: 50 30 70 20 40 60 80


Output:

20 30 40 50 60 70 80

```python
class Node:
    def __init__(self,data):
        self.right=None
        self.left=None
        self.data=data
def insert(root,key):
    if root is None:
        return Node(key)
    if key<root.data:
        root.left=insert(root.left,key)
    else:
        root.right=insert(root.right,key)
    return root
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data,end=" ")
        inorder(root.right)
elements=[50,30,70,20,40,60,80]
root=None
for value in elements:
    root=insert(root,value)
inorder(root)                
```

QUESTION 6: BFS + Path Existence

Topics: BFS + Path Checking

Task:
Using BFS, check whether a path exists between two given nodes.

Input:

graph = {
    'A': ['B'],
    'B': ['C'],
    'C': ['D'],
    'D': []
}
start = 'A'
end = 'D'


Output:

Path Exists

```python
from collections import deque

def path_existence_bfs(graph,start,end):
    visited=set()
    queue=deque([start])
    while queue:
        node=queue.popleft()
        if node==end:
            print("Path Existence")
            return
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    print("Path don't exist")            
graph = {
    'A': ['B'],
    'B': ['C'],
    'C': ['D'],
    'D': []
}
start = 'A'
end = 'D'
path_existence_bfs(graph, start, end)
```

QUESTION 7: DFS + Count Connected Components

Topics: DFS + Connected Graph

Task:
Count the number of connected components in an undirected graph.

Input:

graph = {
    0: [1],
    1: [0],
    2: [3],
    3: [2],
    4: []
}


Output:

Connected Components: 3

```python
def dfs(graph,node,visited):
    visited.add(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph,neighbor,visited)
def count_connected_components(graph):
    visited=set()
    count=0
    for node in graph:
        if node not in visited:
            dfs(graph,node,visited)
            count+=1
    print("Connected components: ",count)
graph = {
    0: [1],
    1: [0],
    2: [3],
    3: [2],
    4: []
}
count_connected_components(graph)                            
```

QUESTION 8: BST + Search

Topics: Binary Search Tree + Search

Task:
Create a BST and search for a given value.

Input:

Elements: 10 5 15 3 7
Search: 7


Output:

Element Found

```python
class Node:
    def __init__(self,data):
        self.right=None
        self.left=None
        self.data=data
def insert(root,key):
    if root is None:
        return Node(key)
    if key<root.data:
        root.left=insert(root.left,key)
    else:
        root.right=insert(root.right,key)
    return root
def search(root,key):
    if root is None:
        return False
    if key==root.data:
        return True
    elif key<root.data:
        return search(root.left,key)
    else:
        return search(root.right,key)
elements=[10,5,15,3,7]
search_value=7

root=None
for value in elements:
    root=insert(root,value)

if search(root,search_value):
    print("Element Found")
else:
    print("Element Not Found")                    
```
